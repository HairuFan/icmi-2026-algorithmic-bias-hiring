Then place the three cleaned CSV files in `data/`.

````python
from pathlib import Path
import sys

# Works whether Jupyter starts in the project root or notebooks/
PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

sys.path.insert(0, str(PROJECT_ROOT / "src"))

from icmi_bias_analysis import run_analysis

DATA_DIR = PROJECT_ROOT / "data"
OUT_DIR = PROJECT_ROOT / "outputs"

ADULT_PATH = DATA_DIR / "Adult_Income_Cleaned.csv"
COMPAS_PATH = DATA_DIR / "COMPAS_Cleaned.csv"
SALARY_PATH = DATA_DIR / "Job_Salary_Cleaned.csv"

ADULT_PATH, COMPAS_PATH, SALARY_PATH, OUT_DIR
````

## 2. Run the complete analysis

````python
results = run_analysis(
    ADULT_PATH,
    COMPAS_PATH,
    SALARY_PATH,
    OUT_DIR,
    test_size=0.30,
    random_state=42,
    bootstrap_B=2000,
)

print("Analysis complete.")
print("Outputs:", OUT_DIR)
````
